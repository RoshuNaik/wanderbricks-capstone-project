```sql
use databricks_wanderbricks_dataset_dais_2025.wanderbricks;


select * from hosts limit 10;

-- 1. Total count of hosts
select count(distinct host_id) as total_hosts
from hosts;


-- 2. Country wise hosts
select country, count(*) as total_hosts
from hosts
group by country
order by total_hosts desc;




--- 3. Verified Vs Not - Verified
select is_verified,COUNT(host_id) AS total_hosts
from hosts
group by is_verified;

--- 4. Active Vs Non- active
select case when is_active = true then 'Active'else 'Non-Active'
    END AS activity_status,
    COUNT(host_id) AS total_hosts
FROM hosts
GROUP BY is_active;

--- 5. By host rating

select case
when rating >= 4.5 then "4.5+"
when rating >= 4 then "4.0 - 4.49"
when rating >= 3.5 then "3.5 - 3.99"
else 'Below 3.5'
end as rating_range,
count(*) as total_hosts
from hosts
group by rating_range
order by
case
when rating_range = '4.5+' then 1
when rating_range = '4.0 - 4.49' then 2
when rating_range = '3.5 - 3.99' then 3
else 4
end;










--- 6. When did host joined the platform

SELECT 
year(joined_at) as joined_year, 
month(joined_at) as month_no,
date_format(joined_at,"MMM") AS joined_month,
COUNT(host_id) AS total_hosts
FROM hosts
GROUP BY 
year(joined_at),
month(joined_at),
date_format(joined_at,"MMM")
ORDER BY joined_year,month_no;

select min(joined_at) as first,max(joined_at) as latest from hosts;

--- 7. How many properties does each host manage.
select h.host_id, h.name,count(p.property_id) as total_properties
from hosts h
left join properties p
on h.host_id = p.host_id
where h.is_active = true
group by h.host_id, h.name
having count(p.property_id) > 0
order by  total_properties desc ;

select * from hosts limit 10;



--- 7. How many properties does each host manage. (label)

select
case
when total_properties >= 10 then '10+ Properties'
when total_properties between 7 and 9  then '7-9 Properties'
when total_properties between 4 and 6  then '4-6 Properties'
when total_properties between 2 and 3 then  '2-3 Properties'
else '1 Property'
end as property_range,
count(*) as total_hosts
from 
(
    select h.host_id, count(*) as total_properties
    from hosts h
    join properties p
    on h.host_id = p.host_id
    where h.is_active = true
    group by h.host_id
)
group by property_range
order by 
case 
when property_range = '10+ Properties' then 1
when property_range = '7-9 Properties' then 2
when property_range = '4-6 Properties' then 3
when property_range = '2-3 Properties' then 4
else 5
end;

select count(*) as hp
from 
( select h.host_id, count(*) as total_properties
    from hosts h
    join properties p
    on h.host_id = p.host_id
    where h.is_active = true
    group by h.host_id
    having count(p.property_id) = 1) x;





```
