```sql
use databricks_wanderbricks_dataset_dais_2025.wanderbricks;

show tables;
select * from bookings limit 10;

-- 1. Size and shape

describe  bookings;

-- total booking record

select count(*) from bookings;

select count(distinct(user_id)) from bookings;

select status, count(booking_id) from bookings group by status;



-- 2. Actual bookings

select count(*) as actual_bookings from bookings where status = 'confirmed' or status = 'completed';

-- 3. How many properties received bookings
select count(distinct property_id) as booked_properties
from bookings 
where status = 'confirmed' or status = 'completed';

-- 4. total revenue by property type.
 
-- 5. seasonal trends

select year(check_in) as year,
month(check_in) as month,
count(*) as total_bookings
from bookings
group by year(check_in),  month(check_in)
order by year,month;

select month(check_in) as month, status, count(*) as total_bookings 
from bookings
group by month(check_in), status
order by month, status;


select date_format(check_in, 'yyyy-MM') as month,
sum(case when status = 'completed' then 1 else 0 end) as completed,
sum(case when status = 'cancelled' then 1 else 0 end) as cancelled,
sum(case when status = 'pending' then 1 else 0 end) as pending,
sum(case when status = 'confirmed' then 1 else 0 end) as confirmed
from bookings
group by date_format(check_in, 'yyyy-MM')
order by month;

-- June to August 2023 shows the strongest booking activity.
-- Dec 2023 also shows another spike.
-- Jan-May and Sep-Nov are comparatively lower.


-- 6. booking status

-- 7. what time period does booking data cover?

-- 8. stay duration

-- 9. avg guest count by property type/ seasons. 

-- 10. booking amount

-- 11. repeat booking behaviour

select * from bookings;

select user_id, count(*) as total_count from bookings group by user_id order by total_count desc;

select total_bookings, count(*) as users
from
(select user_id, count(*) as total_bookings from bookings group by user_id)
group by total_bookings
order by total_bookings;

-- 1. Most users have booked more than once.
-- 2. Number of users decreases steadily as booking frequency increases.
-- 3. Very few users have made more than 8 bookings

-- 12. property performance


select p.title, count(*) as total_bookings,
sum(case when b.status = 'completed' then 1 else 0 end) as completed_booking,
sum(case when b.status = 'cancelled' then 1 else 0 end) as cancelled_booking,
round(sum(case when b.status = 'completed' then total_amount else 0 end),2) as revenue
from bookings b
join properties p
on p.property_id = b.property_id
group by  p.title
order by revenue desc
limit 10;



-- Serviced Residence in Singapore generated the highest completed booking revenue.
-- Hotel in Singapore and Apartment are also among highest revenue generating properties.
-- For many of the top property cancelled booking are higher.



show tables;

```
